```python
df["imf_program_active"] = df["imf_program_active"].astype(bool)

df["decade"] = df["decade"].astype("category")

df["gdp_growth_category"] = df["gdp_growth_category"].astype("category")
df["inflation_category"]  = df["inflation_category"].astype("category")

print("\nUpdated dtypes:")
print(df.dtypes)
```

```python
# Real GDP per capita growth: how much richer did the average person get?
# GDP growth minus population growth approximates this
df["population_growth_pct"] = df["population_mn"].pct_change() * 100
df["real_gdp_per_capita_growth"] = df["gdp_growth_pct"] - df["population_growth_pct"]

# Economic stress score: combines inflation and unemployment into one signal
# Simple unweighted sum — higher means more pressure on ordinary people
def stress_score(row):
    return round(row["inflation_cpi_pct"] + row["unemployment_pct"], 2)

df["economic_stress_score"] = df.apply(stress_score, axis=1)

df[["year", "gdp_growth_pct", "real_gdp_per_capita_growth",
    "inflation_cpi_pct", "unemployment_pct", "economic_stress_score"]].head(10)
```

```python
summary = df.pivot_table(
    index="decade",
    columns="imf_program_active",
    values=["gdp_growth_pct", "inflation_cpi_pct", "economic_stress_score"],
    aggfunc="mean"
).round(2)


summary
```

```python
decade_context = pd.DataFrame({
    "decade": ["2000s", "2010s", "2020s"],
    "dominant_theme": [
        "Post-sanctions recovery + 9/11 dividend",
        "CPEC boom + twin deficit buildup",
        "COVID + PKR crash + IMF dependency"
    ],
    "overall_rating": ["Improving", "Mixed", "Crisis"]
})

df_enriched = df.merge(decade_context, on="decade", how="left")

df_enriched[["year", "decade", "gdp_growth_pct", "dominant_theme", "overall_rating"]].head(12)
```
